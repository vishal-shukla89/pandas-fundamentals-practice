```python
import numpy as np 
import pandas as pd
```

### **(1)Creating a DataFrame**

```python
#there are two ways to create dataframe by using multiple lists and by using Dictionary
```

```python
#(a)Creating dataframe by using dictionary
data={
    'Name':['Vishal','Aman','Nikhil','Sumit'],
    'Age':[23,23,21,22],
    'City':['Mumbai','New york','Paris','London'],
    'Salary':[89000,60000,80000,81000]
}
df=pd.DataFrame(data)
df
```

```python
#(b)Creating data frame by using multuple lists 
my_list=[
    ['Vishal',23,'Mumbai',89000],
    ['Aman',23,'New york',60000],
    ['Nikhil',21,'Paris',80000],
    ['Sumit',22,'London',81000]
] 
df2=pd.DataFrame(my_list)
df2   #we got the dataframe but with deafult index and column name 
```

```python
#here we will create two more list for index and column name and then set them into dataframe
ind=[1,2,3,4]
col_names=['Name','Age','City','Monthly Savings']
df2=pd.DataFrame(my_list,index=ind,columns=col_names)
df2
```

### **(2)Selection And Indexing of Columns**

```python
df2
```

```python
#for selecting any one column 
df2['City']
```

```python
#for selecting multiple columns
df2[['Age','City']]
```

### **(3)Creating new Column inside dataframe**

```python
#let we want add a column for designation of each person
df2['Designation']=['Data Scientist','unkown','Data Scientist','Engeneer']
df2
```

### **(4)Removing Columns**

```python
#for removing any column we use .drop(col_name,index=0/1,inplace=True/False)
#if we put only column name then it will throw error as defalt axis is zero=row wise remove
#if we dont write inplace=True then it will just show an example output not change in original datafarme
```

```python
#exaple1
df2.drop('Designation',axis=1)  #axis=1 for column and axis=0 for row
```

```python
df2  #as you can see in original dataframe desihgnation not removed
```

```python
#example2
df2.drop('Designation',axis=1,inplace=True)  #now this will permanently delete the column
df2
```

```python
#example3
df2.drop(["Age","City"],axis=1,inplace=True)  #for deleting multiple columns
df2
```

### **(6)Selection of subset of rows and column**

```python
#example1- 1:2Bycity:salary
df1.loc[[1,2]][['City','Salary']]
```

```python
#example2-want row=2,4 and col=name, city, salary
df1.loc[[2,4]][['Name',"City",'Salary']]
```

### **(7)Conditional Selection**

```python
#for one condition
#example- give the data of those people whoose age is greater than 21
df1[df1['Age']>21]
```

```python
#For multiple condition
#example2-1st condition and city must be London
df1[(df1['Age']>21) & (df1['City']=='London')]
```
