```python
import numpy as np 
import pandas as pd
```

### **(A)Finding Missing Data**

```python
data={
    'A':[1,2,3,np.nan,5],
    'B':[np.nan,2,3,4,6],
    'C':[2,3,5,4,7],
    'D':[np.nan,np.nan,2,3,np.nan]
}
df=pd.DataFrame(data)
df
```

```python
#to find missing data we use .isna()
df.isna()
```

```python
#as above you can see the null values gives TRUE
#now we want to calculte how many null values are there in each column
#for this sort data counting manually is possible but what about for 1000 rows
#for this we use .sum() after .isna()
df.isna().sum()
```

```python
#if want to know if there are any blank cell or any type of null values in each column then
#we use .any()
df.isna().any()
```

### **(B)Removing missing Data**

```python
df
```

```python
#for removing null values we use .dropna() but it removes whole row if it contains any null values
#but we can use threshold to to control the removal 
#likr threshold=3 (means it will remove all the row which has less than 3 real values)

#example1-without threshold
df.dropna()
```

```python
#example2-with threshold
df.dropna(thresh=3)
```

```python
#as above you can see only first row got removed as thresh=3
#all above things didnt change original dataframe as we had not added inplace=TRUE
```

### **(C)Filling The Missing DATA**

```python
df
```

```python
#if we want to replace na values to any number we can by this way
df.fillna(0)  #but this is only for display if really want to change in data put inplace=TRUE
```

```python
#If we want to fill column wise different values we can do by this way
values={'A':100,'B':200,'C':300,'D':400}
df.fillna(value=values)
```

```python
#we also can replce na values with the mean of respective column
df.fillna(df.mean())
```
