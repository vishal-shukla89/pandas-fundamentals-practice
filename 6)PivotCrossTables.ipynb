### **(A)PIVOT TABLE**
*-here we use datframe,index/rows=col1(it should be categorical),columns=col2(it should be categorical),values=aggfun*

*-now in pivot table for values it takes default aggregation function = mean of all values that are coming for each particular cell of table*

*-values must be numeric*

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}
df=pd.DataFrame(data)
```

```python
#we want Date in more specified way like month specified and quarter as well
df['Month']=df['Date'].dt.month_name()
df['Quarter']='Q'+df['Date'].dt.quarter.astype(str)
```

```python
df
```

**Q.1 Create a Pivot table with rows as "Region" and columns as "Product" with value of each cell as "Sales"**

```python
#1st direct way and default aggregation function is mean thats why all the values are mean 
df.pivot_table(values='Sales',index='Region',columns='Product')
```

```python
#2nd indirect way and here we'll take aggregation function as max
pd.pivot_table(df,values='Sales',index='Region',columns='Product',aggfunc='max')
```

**Q.2 same as previous question just for values include one more data as 'Units'**

```python
pivot_2=df.pivot_table(values=['Sales','Units'],index='Region',columns='Product')
pivot_2
```

**Q.1 Create a cross table with rows as 'Region' and columns as 'Product'**
