### **(A)GroupBy**
*-this is used to group categorical data* 

*-like a column contains A and B types of category multiple times*

*-then by using groupby() we can group A and B differently and perfom tasks*

```python
data={
    'Category':['A','B','A','B','A','B','A','B'],
    'Store':['S1','S1','S2','S2','S1','S2','S2','S1'],
    'Sales':[100,200,150,250,120,180,200,300],
    'Quantity':[10,15,12,18,8,20,15,25],
    'Date':pd.date_range('2023-01-01',periods=8)
}
df=pd.DataFrame(data)
df
```

**Q.1 Group by category and calculte the sum of sales**


```python
cat=df.groupby('Category')
cat
```

```python
#since by groupby it creates object which we cants see but by loop we can see
for i,j in cat:
    print(i)
    print(j)
```

```python
#now for sum of sales
df.groupby('Category')['Sales'].sum()
```

**Q.2 Group by store and calculate total quantity**

```python
df.groupby('Store')['Quantity'].sum()
```

**Q.3 Group by multiple columns like category and store both then shoe sum of sales**

```python
df.groupby(['Category','Store'])['Sales'].sum()
```

### **(B)Aggregation**
*-means combining all data of particular column*

*-then can use aggregation functions on them*

*-like mean,median,variance,std,min,max,count*

**Q.1 perform aggregation function means on sales column**

```python
df['Sales'].mean()
```

**Q.2 by only one line show all aggregation functions use like mean,median,variance,std,min,max,count**

```python
df['Sales'].agg(['mean','median','var','std','min','max','count'])
```
