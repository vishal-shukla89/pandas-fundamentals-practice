### **(A)SOME BASIC DATAFRAME OPERATIONS**
*-here we are going to show basic methods as shape ,columns*

*-also some functions as like .info(),.describe()*

```python
df1=pd.DataFrame({
    'A':[1,2,3,4,5],
    'B':[10,20,30,40,50],
    'C':[100,200,300,400,500]
})
df1
```

```python
#by boradcasting add 10 to B column
df1['B']=df1['B']+10
df1
```

### **(B)APPLY OPERATION**
*- used as .apply(fun_) and by using this we can apply any function to change values of any column*

**Q.change column B by squres of them**

```python
#Basic use like by defining an function
def square(x):
    return x**2

df1['B'].apply(square)
```

```python
df1
```

```python
#as above u can see that in original dataframe changes not taken placed
#below is an example how to do it
df1['B']=df1['B'].apply(square)
df1
```

```python
#advance way is by using lambda function instead of defining function , but now we will do A
df1['A']=df1['A'].apply(lambda x:x**2)
df1
```
