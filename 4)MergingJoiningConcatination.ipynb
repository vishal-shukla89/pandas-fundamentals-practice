### **(A)Merging 2 DataFrames**

```python
#DataFrame 1 for Employee details
employees=pd.DataFrame({
    'employee_id':[1,2,3,4,5],
    'Name':['Vishal','Sumit','Vipin','Abhishek','Avs'],
    'department':['IT','HR','Finance','HR','IT']
}
)
#DataFrame 2 for SAlary information
salaries=pd.DataFrame({
    'employee_id':[1,2,3,6,7],
    'salary':[89000,82000,70000,80000,72000],
    'bonus':[10000,5000,8000,1000,6000]
})
```

```python
employees
```

```python
salaries
```

```python
#now we want to show dataframe by combining both dataframe 
#in different ways
```

```python
#show on valid valuse by combining both dataframes 
#we use two important parameters (df1,df2,....,on= ,how= )
pd.merge(employees,salaries)  #this is default 
                              #where on=matching column of both dataframe and
                              #how='inner' means only those rows which are intersection of both dataframes
```

```python
pd.merge(employees,salaries,on='employee_id',how='inner') #now this is with how='inner'
```

```python
#using how='outer'
#we are merging both Dataframes i.e..,unioin(showing all rows of both dataframes)
pd.merge(employees,salaries,on='employee_id',how='outer')  
```

```python
#by using how='left'
#we are giving more preference to left dataframe i.e.., To employees
#means will show all the rows of left datafarme but including common entries based on emploee_id
pd.merge(employees,salaries,on='employee_id',how='left')
```

```python
#by using how='right'
#we are giving more preference to right dataframe i.e.., To salaries
#means will show all the rows of right datafarme but including common entries based on emploee_id

pd.merge(employees,salaries,on='employee_id',how='right')
```

### **(B)Concatination Of 2 DataFrames**

```python
df1=pd.DataFrame({
    'A':['A0','A1','A2'],
    'B':['B0','B1','B2'],
    'C':['C0','C1','C2']
})
df2=pd.DataFrame({
    'A':['A3','A4','A5'],
    'B':['B3','B4','B5'],
    'C':['C3','C4','C5']
})
```

```python
df1
```

```python
df2
```

```python
#Concatination means pasting dataframes horizontally or vertically 
#(like combining both beside or up to down)
```

```python
#axis=0 is an default parameter
#by using this we can paste other datafarme vertically(row wise i.e.., below to first dataframe)
pd.concat([df1,df2],axis=0) 
```

```python
#axis=1
#by using this we can paste other datafarme horizontalli(column wise i.e.., beside to first dataframe)
pd.concat([df1,df2],axis=1) 
```

### **(C)Joining 2 DataFrames**

```python
df1=pd.DataFrame({
    'name':['Vishal','Sumit','Vipin']
},index=[1,2,3])
df1
```

```python
df2=pd.DataFrame({
    'score':[85,80,90]
},index=[2,3,4])
df2
```

```python
#joining two dataframes means joining one dataframes column to other dataframe
#but here also use of how='inner' or how='outer'
```

```python
#by how='inner' means intesection of both dataframe based on common entries 
#here common entries are index
df1.join(df2,how='inner')
```

```python
#by how='outer' means union of both dataframe based on common entries 
#here common entries are index
df1.join(df2,how='outer')
```
