### **Q.10 How many countries are there who has republic in there long name**

```python
#first of all we'll make an function which directly counts the country with republic in their long name
count=0
def counting(txt):
    global count
    if 'republic' in txt.lower():
        count+=1
    return txt

df['country_long']==df['country_long'].apply(counting)
print(count)
```
