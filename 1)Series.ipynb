### PNADAS SERIES
A Pandas Series is a one dimensional labeled array capable of holding any data type.
    the axis labels are collectively called index.

### **Creating Pandas series**
there are multiple types of way

```python
#we will create series using different data structures
import numpy as np 
import pandas as pd
```

```python
labels=['a','b','c','d'] #list with strings
my_list=[10,20,30,40]  #list with integers 
arr=np.array([10,20,30])
d={'a':10,'b':20,'c':30,'d':40}
```

```python
pd.Series(labels) #series always comes in columns
```

```python
pd.Series(my_list) #for list to series
```

```python
pd.Series(my_list,index=labels)  #here we are changing default index to a,b,c,d
```

```python
pd.Series(arr)
```

```python
pd.Series(d)  #in dictionary our keys get saved as index and values as series
```
